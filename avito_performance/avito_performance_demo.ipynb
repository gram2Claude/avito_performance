## 2. Импорты и загрузка учётных данных

```python
import logging
import os
import sys
from dotenv import load_dotenv

sys.path.insert(0, ".")  # avito_performance.py лежит рядом с notebook
load_dotenv(".env")
logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s")

assert os.environ.get("CLIENT_ID"), "CLIENT_ID отсутствует в .env"
assert os.environ.get("CLIENT_SECRET"), "CLIENT_SECRET отсутствует в .env"
assert os.environ.get("ACCOUNT_ID"), "ACCOUNT_ID отсутствует в .env"

print("Учётные данные загружены")
```

```python
from avito_performance import (
    get_campaign_dict,
    get_group_dict,
    get_creative_dict,
    get_campaigns_daily_stat,
    get_groups_daily_stat,
    get_creatives_daily_stat,
)
```

## 3. Параметры периода

```python
from datetime import date, timedelta

DATE_TO   = (date.today() - timedelta(days=1)).isoformat()
DATE_FROM = (date.today() - timedelta(days=7)).isoformat()

# Переопредели вручную если нужно (период ≤ 100 дней):
# DATE_FROM = "2026-01-01"
# DATE_TO   = "2026-01-31"

print(f"Период: {DATE_FROM} → {DATE_TO}")
```

## 4. Справочник кампаний — `get_campaign_dict()`

```python
get_campaign_dict_result = get_campaign_dict()
print(f"Строк: {len(get_campaign_dict_result)}")
get_campaign_dict_result.head(10)
```

## 5. Справочник групп — `get_group_dict()`

```python
get_group_dict_result = get_group_dict()
print(f"Строк: {len(get_group_dict_result)}")
get_group_dict_result.head(10)
```

## 6. Справочник креативов — `get_creative_dict()`

```python
get_creative_dict_result = get_creative_dict()
print(f"Строк: {len(get_creative_dict_result)}")
get_creative_dict_result.head(10)
```

## 7. Статистика по кампаниям — `get_campaigns_daily_stat(date_from, date_to)`

```python
get_campaigns_daily_stat_result = get_campaigns_daily_stat(DATE_FROM, DATE_TO)
print(f"Строк: {len(get_campaigns_daily_stat_result)}")
get_campaigns_daily_stat_result.head(10)
```

```python
# Сводка по дням
if not get_campaigns_daily_stat_result.empty:
    display(get_campaigns_daily_stat_result.groupby("date")[["views", "clicks", "costs_nds", "costs_without_nds", "costs_nds_ak", "costs_without_nds_ak"]].sum())
```

## 8. Статистика по группам — `get_groups_daily_stat(date_from, date_to)`

```python
get_groups_daily_stat_result = get_groups_daily_stat(DATE_FROM, DATE_TO)
print(f"Строк: {len(get_groups_daily_stat_result)}")
get_groups_daily_stat_result.head(10)
```

```python
# Сводка по дням
if not get_groups_daily_stat_result.empty:
    display(get_groups_daily_stat_result.groupby("date")[["views", "clicks", "costs_nds", "costs_without_nds", "costs_nds_ak", "costs_without_nds_ak"]].sum())
```

## 9. Статистика по креативам — `get_creatives_daily_stat(date_from, date_to)`

```python
get_creatives_daily_stat_result = get_creatives_daily_stat(DATE_FROM, DATE_TO)
print(f"Строк: {len(get_creatives_daily_stat_result)}")
get_creatives_daily_stat_result.head(10)
```

```python
# Сводка по дням
if not get_creatives_daily_stat_result.empty:
    display(get_creatives_daily_stat_result.groupby("date")[["views", "clicks", "costs_nds", "costs_without_nds", "costs_nds_ak", "costs_without_nds_ak"]].sum())
```

## Сохранение в CSV (опционально)

```python
from datetime import date

today = date.today().isoformat()

get_campaign_dict_result.to_csv(f"get_campaign_dict_{today}.csv", index=False, encoding="cp1251", errors="replace")
get_group_dict_result.to_csv(f"get_group_dict_{today}.csv", index=False, encoding="cp1251", errors="replace")
get_creative_dict_result.to_csv(f"get_creative_dict_{today}.csv", index=False, encoding="cp1251", errors="replace")
get_campaigns_daily_stat_result.to_csv(f"get_campaigns_daily_stat_{DATE_FROM}_{DATE_TO}.csv", index=False, encoding="cp1251", errors="replace")
get_groups_daily_stat_result.to_csv(f"get_groups_daily_stat_{DATE_FROM}_{DATE_TO}.csv", index=False, encoding="cp1251", errors="replace")
get_creatives_daily_stat_result.to_csv(f"get_creatives_daily_stat_{DATE_FROM}_{DATE_TO}.csv", index=False, encoding="cp1251", errors="replace")

print("Файлы сохранены")
```
